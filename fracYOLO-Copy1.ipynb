```python
from ultralytics import YOLO
```

```python
model = YOLO('best.pt')
```

```python
results = model.train(
    data="data.yaml",
    epochs=500, 
    imgsz=1024, 
    batch=16,
    single_cls=True, 
    mode='train', 
    task='detect', 
    device=[1,2,3],
    freeze=5,
    mosaic=0.9,
    lr0=0.001,
    close_mosaic=50, 
    patience=150
)
```
