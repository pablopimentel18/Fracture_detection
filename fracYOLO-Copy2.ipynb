```python
from ultralytics import YOLO
```

```python
model = YOLO('best.pt')
```

```python
results = model.train(
    data="data.yaml",   
    mode='train',
    task='detect',
    epochs=500,     
    batch=8,             
    imgsz=1280,       
    save=True,
    device=[1,2,5,6],
    pretrained=True,
    patience=150,           
    plots=True, 
    exist_ok=True,
    lr0=0.001,
    single_cls=True,
    freeze=10,
    degrees=0.9,
)
```

```python
from ultralytics import YOLO
model = YOLO('runs/detect/train/weights/best.pt')

metrics = model.val()
print(metrics)
```
